```python
import pandas as pd
import numpy as np
import json

# Combine training and validation frames
all_frames = training_frames.union(validation_frames)

# Initialize lists to store data
weather_conditions = []
road_conditions = []
time_of_day = []
road_types = []
num_vehicles = []
num_lane_instances = []
num_pedestrians = []
num_traffic_lights = []
num_traffic_signs = []
speeds_kph = []

# New columns
frame_ids = []
country_codes = []
longitudes = []
latitudes = []
times = []

# Loop through all frames
for frame_id in all_frames:
    zod_frame = zod_frames[frame_id]

    try:
        # Extract features
        weather_condition = zod_frame.metadata.scraped_weather
        road_condition = zod_frame.metadata.road_condition
        time_day = zod_frame.metadata.time_of_day
        road_type = zod_frame.metadata.road_type
        num_vehicle = zod_frame.metadata.num_vehicles
        num_lane_instance = zod_frame.metadata.num_lane_instances
        num_pedestrian = zod_frame.metadata.num_pedestrians
        num_traffic_light = zod_frame.metadata.num_traffic_lights
        num_traffic_sign = zod_frame.metadata.num_traffic_signs

        # Extract new features
        frame_id = zod_frame.metadata.frame_id
        country_code = zod_frame.metadata.country_code
        longitude = zod_frame.metadata.longitude
        latitude = zod_frame.metadata.latitude
        time = zod_frame.metadata.time

        # Calculate speed (assuming one speed per frame)
        Vx = zod_frame.ego_motion.velocities[0, 0]
        Vy = zod_frame.ego_motion.velocities[0, 1]
        Vz = zod_frame.ego_motion.velocities[0, 2]
        speed_mps = np.sqrt(Vx**2 + Vy**2 + Vz**2)
        speed_kph = speed_mps * 3.6

        # Append to lists
        weather_conditions.append(weather_condition)
        road_conditions.append(road_condition)
        time_of_day.append(time_day)
        road_types.append(road_type)
        num_vehicles.append(num_vehicle)
        num_lane_instances.append(num_lane_instance)
        num_pedestrians.append(num_pedestrian)
        num_traffic_lights.append(num_traffic_light)
        num_traffic_signs.append(num_traffic_sign)
        speeds_kph.append(speed_kph)

        # Append new features
        frame_ids.append(frame_id)
        country_codes.append(country_code)
        longitudes.append(longitude)
        latitudes.append(latitude)
        times.append(time)

    except json.JSONDecodeError as e:
        print(f"Skipping frame {frame_id} due to JSONDecodeError: {e}")
    except AttributeError as e:
        print(f"Skipping frame {frame_id} due to missing attribute: {e}")

# Create a DataFrame to analyze the data
data = pd.DataFrame({
    'Frame ID': frame_ids,
    'Country Code': country_codes,
    'Longitude': longitudes,
    'Latitude': latitudes,
    'Time': times,
    'Weather Condition': weather_conditions,
    'Road Condition': road_conditions,
    'Time of Day': time_of_day,
    'Road Type': road_types,
    'Number of Vehicles': num_vehicles,
    'Number of Lane Instances': num_lane_instances,
    'Number of Pedestrians': num_pedestrians,
    'Number of Traffic Lights': num_traffic_lights,
    'Number of Traffic Signs': num_traffic_signs,
    'Speed (kph)': speeds_kph
})

# Display the first few rows of the dataframe to verify
print(data.head())

# Save to CSV file
data.to_csv('data.csv', index=False)

```

```python
import os
import json
import numpy as np
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt

base_path = '/content/drive/My Drive/Colab Notebooks/ZODMiniDataset/Frames/single_frames'

# Function to get lane marking data
def get_lane_marking_data(frame_id):
    annotation_path = os.path.join(base_path, frame_id, 'annotations', 'lane_markings.json')
    with open(annotation_path, 'r') as file:
        lane_markings = json.load(file)
    return lane_markings

# Function to flatten the coordinates
def flatten_coordinates(coords):
    return [tuple(coord) for coord in coords]

# Function to convert lane markings to binary mask
def convert_lane_markings_to_mask(lane_markings, image_size):
    mask = Image.new('L', image_size, 0)
    draw = ImageDraw.Draw(mask)

    for marking in lane_markings:
        polygon = marking['geometry']['coordinates'][0]
        flattened_polygon = flatten_coordinates(polygon)
        draw.polygon(flattened_polygon, outline=1, fill=1)

    return np.array(mask)

# Function to extract country code from metadata
def get_country_code(frame_id):
    metadata_path = os.path.join(base_path, frame_id, 'metadata.json')
    with open(metadata_path, 'r') as file:
        metadata = json.load(file)
    return metadata['country_code']

# Function to find the image file
def find_image_file(frame_id):
    frame_path = os.path.join(base_path, frame_id, 'camera_front_blur')
    for file_name in os.listdir(frame_path):
        if file_name.endswith('.jpg'):
            return os.path.join(frame_path, file_name)
    return None

# Example usage for selected frames
frame_ids = ['007674', '009158', '018591', '023996', '029229', '044953', '062592', '065838', '070221', '082291', '083430', '087912']

for frame_id in frame_ids:
    image_path = find_image_file(frame_id)
    if image_path:
        image = Image.open(image_path)
        lane_markings = get_lane_marking_data(frame_id)
        mask = convert_lane_markings_to_mask(lane_markings, image.size)

        plt.figure()
        plt.subplot(1, 2, 1)
        plt.imshow(image)
        plt.title(f'Image {frame_id}')

        plt.subplot(1, 2, 2)
        plt.imshow(mask, cmap='gray')
        plt.title(f'Lane Markings Mask {frame_id}')

        plt.show()
    else:
        print(f'Image file not found for frame {frame_id}')
```

```python
import segmentation_models_pytorch as smp
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import os
import json
import numpy as np
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt

base_path = '/content/drive/My Drive/Colab Notebooks/ZODMiniDataset/Frames/single_frames'

class LaneMarkingDataset(Dataset):
    def __init__(self, base_path, frame_ids, transform=None):
        self.base_path = base_path
        self.frame_ids = frame_ids
        self.transform = transform

    def __len__(self):
        return len(self.frame_ids)

    def __getitem__(self, idx):
        frame_id = self.frame_ids[idx]
        image_path = find_image_file(frame_id)
        if image_path:
            image = Image.open(image_path)

            lane_markings = get_lane_marking_data(frame_id)
            mask = convert_lane_markings_to_mask(lane_markings, image.size)

            if self.transform:
                image = self.transform(image)
                mask = Image.fromarray(mask)  # Convert mask to PIL image
                mask = self.transform(mask)

            return image, mask
        else:
            raise FileNotFoundError(f'Image file not found for frame {frame_id}')

def get_lane_marking_data(frame_id):
    annotation_path = os.path.join(base_path, frame_id, 'annotations', 'lane_markings.json')
    with open(annotation_path, 'r') as file:
        lane_markings = json.load(file)
    return lane_markings

def flatten_coordinates(coords):
    return [tuple(coord) for coord in coords]

def convert_lane_markings_to_mask(lane_markings, image_size):
    mask = Image.new('L', image_size, 0)
    draw = ImageDraw.Draw(mask)

    for marking in lane_markings:
        polygon = marking['geometry']['coordinates'][0]
        flattened_polygon = flatten_coordinates(polygon)
        draw.polygon(flattened_polygon, outline=1, fill=1)

    return np.array(mask)

def get_country_code(frame_id):
    metadata_path = os.path.join(base_path, frame_id, 'metadata.json')
    with open(metadata_path, 'r') as file:
        metadata = json.load(file)
    return metadata['country_code']

def find_image_file(frame_id):
    frame_path = os.path.join(base_path, frame_id, 'camera_front_blur')
    for file_name in os.listdir(frame_path):
        if file_name.endswith('.jpg'):
            return os.path.join(frame_path, file_name)
    return None

frame_ids = ['007674', '009158', '018591', '023996', '029229', '044953', '062592', '065838', '070221', '082291', '083430', '087912']

transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.ToTensor()
])

dataset = LaneMarkingDataset(base_path, frame_ids, transform)
dataloader = DataLoader(dataset, batch_size=4, shuffle=True)

model = smp.Unet('resnet34', classes=1, activation='sigmoid').cuda()

criterion = torch.nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

for epoch in range(5):
    model.train()
    for images, masks in dataloader:
        images, masks = images.cuda(), masks.cuda()

        outputs = model(images)
        loss = criterion(outputs, masks)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print(f'Epoch {epoch+1}, Loss: {loss.item()}')

model.eval()
with torch.no_grad():
    for images, masks in dataloader:
        images, masks = images.cuda(), masks.cuda()
        outputs = model(images)

        images_np = images.cpu().numpy().transpose(0, 2, 3, 1)
        masks_np = masks.cpu().numpy()
        outputs_np = outputs.cpu().numpy()

        for i in range(images_np.shape[0]):
            plt.figure()
            plt.subplot(1, 3, 1)
            plt.imshow(images_np[i])
            plt.title('Image')

            plt.subplot(1, 3, 2)
            plt.imshow(masks_np[i].squeeze(), cmap='gray')
            plt.title('Ground Truth')

            plt.subplot(1, 3, 3)
            plt.imshow((outputs_np[i].squeeze() > 0.5).astype(np.uint8), cmap='gray')
            plt.title('Prediction')

            plt.show()
```

```python
import os
import json
import pandas as pd

base_path = "/content/drive/MyDrive/Colab Notebooks/ZODFullDataset/Frames/single_frames"
data = []

# Function to calculate speed magnitude
def calculate_speed_magnitude(velocities):
    return [(v[0]**2 + v[1]**2 + v[2]**2) ** 0.5 for v in velocities]

# Loop through folders and load data
for folder in sorted(os.listdir(base_path)):
    folder_path = os.path.join(base_path, folder)
    if os.path.isdir(folder_path):
        metadata_path = os.path.join(folder_path, "metadata.json")
        ego_motion_path = os.path.join(folder_path, "ego_motion.json")

        # Check if metadata and ego_motion files exist
        if not os.path.exists(metadata_path) or not os.path.exists(ego_motion_path):
            print(f"Skipping folder {folder} due to missing files.")
            continue

        try:
            with open(metadata_path, 'r') as f:
                metadata = json.load(f)

            with open(ego_motion_path, 'r') as f:
                ego_motion = json.load(f)

            # Calculate speed magnitudes and average speed
            speed_magnitudes = calculate_speed_magnitude(ego_motion['velocities'])
            avg_speed = sum(speed_magnitudes) / len(speed_magnitudes)

            # Calculate total time span
            time_span = max(ego_motion['timestamps']) - min(ego_motion['timestamps'])

            # Append data to list
            data.append({
                "frame_id": metadata["frame_id"],
                "time": metadata["time"],
                "country_code": metadata["country_code"],
                "weather_condition": metadata["scraped_weather"],
                "collection_car": metadata["collection_car"],
                "road_type": metadata["road_type"],
                "road_condition": metadata["road_condition"],
                "time_of_day": metadata["time_of_day"],
                "num_lane_instances": metadata["num_lane_instances"],
                "num_vehicles": metadata["num_vehicles"],
                "num_vulnerable_vehicles": metadata["num_vulnerable_vehicles"],
                "num_pedestrians": metadata["num_pedestrians"],
                "num_traffic_lights": metadata["num_traffic_lights"],
                "num_traffic_signs": metadata["num_traffic_signs"],
                "longitude": metadata["longitude"],
                "latitude": metadata["latitude"],
                "solar_angle_elevation": metadata["solar_angle_elevation"],
                "average_speed": avg_speed,
                "time_span": time_span
            })
        except Exception as e:
            print(f"Error processing folder {folder}: {e}")
            continue

# Convert to DataFrame
df = pd.DataFrame(data)

# Display the first few rows of the DataFrame to verify
df.head()

df.to_csv('vehicle.csv', index=False)
```
